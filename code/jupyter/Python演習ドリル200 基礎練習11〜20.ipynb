       整数を入力し、3と5の公倍数なら「3と5の公倍数です」と表示、
       
       3の倍数なら「3の倍数です」と表示
       
       5の倍数なら「5の倍数です」と表示
       
       どれでもなければ「3の倍数でも5の倍数でもありません」と表示せよ
        
        （実行画面）
```        
    整数を入れてください=15
    3と5の公倍数です
```
```        
    整数を入れてください=9
    3の倍数です
```
```        
    整数を入れてください=10
    5の倍数です
```
```        
    整数を入れてください=8
    3の倍数でも5の倍数でもありません
```

<details><summary>▶プログラム例</summary>

```python
    n = int(input("整数を入れてください="))
    if n % 3 == 0 and n % 5 == 0:
        print("3と5の公倍数です")
    elif n % 3 == 0:
        print("3の倍数です")
    elif n % 5 == 0:
        print("5の倍数です")
    else:
        print("3の倍数でも5の倍数でもありません")
```

    そうでなければ「うるう年ではありません」と表示せよ

<details><summary>▶プログラム例</summary>
    
```python
    year = int(input("西暦を入れてください="))
    if (year % 4 == 0 and year % 100 != 0) or year % 400 == 0:
        print("うるう年です")
    else:
        print("うるう年ではありません")
```

- 課題17

    １から１０の１０個の数字を加算するプログラム
    
    ただし、最初に数字を一つ入力し、その数字は加算しないようにせよ
    
    （実行画面）
    
    ```
    加算しない数字は？5
    合計は50です
    ```
    ```
    加算しない数字は？0
    合計は55です
    ```

<details><summary>▶プログラム例</summary>
    
```python
    n = int(input("加算しない数字は？"))
    total = 0
    for i in range(1, 11):
        if i == n:
            continue
        total += i
    print(f"合計は{total}です")
```

    ただし、5個で改行せよ
    
    （実行画面）
    
    ```
    ＊を何個表示しますか？12
    *****
    *****
    **
    ```

<details><summary>▶プログラム例</summary>
    
```python
    n = int(input("＊を何個表示しますか？"))
    while n > 5:
        print("*****")
        n -= 5
    for i in range(n):
        print("*", end="")
    print()
```

- 課題20

    ランダムに3つの数字を表示させ
    
    3つとも同じ数字だった場合は「Bingo!」と表示
    
    1回実行するごとに、続けるか入力させよ'もう一度（0：はい／1：いいえ）：'
    
    最高10回まで実行
    
    数字を表示する間隔を1秒とせよ
    
    最初の２つの数字が同じ場合、3つ目の表示は5秒後とせよ
    
    （実行画面）
    
    ```
    8 0 6
    もう一度（0：はい／1：いいえ）：0
    1 0 2
    もう一度（0：はい／1：いいえ）：0
    0 1 6
    もう一度（0：はい／1：いいえ）：1
    ```

<details><summary>▶プログラム例</summary>
    
```python
    import random
    import time
    
    for i in range(10):
        n1 = random.randint(0, 9)
        n2 = random.randint(0, 9)
        n3 = random.randint(0, 9)
        print(n1, end=" ")
        time.sleep(1)
        print(n2, end=" ")
        if n1 == n2:
            time.sleep(5)
        else:
            time.sleep(1)
        print(n3)
        if n1 == n2 == n3:
            print("Bingo!")
        if int(input("もう一度（0：はい／1：いいえ）：")) == 1:
            break
```
    
</details>
