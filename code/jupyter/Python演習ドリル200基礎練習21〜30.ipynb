        底辺の長さを入力し、正三角形を表示せよ

- 課題28

    ビット処理　1の数を数える
    
    入力された値（0〜255）を2進数で表したときの1の数を数えよ
    
    次の手順で8桁のビット列の1の数を数えよ
    
    1. 入力した値と00000001との論理積(and)をもとめる（左側7ビットを0にした値）
    2. その値が1ならばカウントする
    3. 入力した値を右に1ビットシフトする
    4. 1〜3を8回繰り返す
    
    （実行画面）
    
    ```
    数を入れて＝198
    11000110
    1の数は4個
    ```

<details><summary>▶プログラム例</summary>
    
```python
    x = int(input("数を入れて＝"))
    print(f"{x:08b}")
    cnt = 0
    for _ in range(8):
        if (x & 0x00000001) == 1:
            cnt += 1
        x >>= 1
    print(f"1の数は{cnt}個")
```
    
</details>

- 課題29

    ビット処理　ビット列をセット
    
    8問の◯☓問題の答えを入力する（◯は1、☓は0）
    
    次の手順で8桁のビット列をセットする
    
    1. 答えを格納する変数xに0を入れる
    2. 入力された答え（0か1）と変数xの論理和(or)を変数xに入れる
    3. 変数xを左に1ビットシフトする
    4. 2〜3を8回繰り返す
    5. 変数xを右に1ビットシフトし、表示する
    
    （実行画面）
    
    ```
    問題 1 （◯は1、☓は0）＝1
    問題 2 （◯は1、☓は0）＝0
    問題 3 （◯は1、☓は0）＝1
    問題 4 （◯は1、☓は0）＝0
    問題 5 （◯は1、☓は0）＝1
    問題 6 （◯は1、☓は0）＝0
    問題 7 （◯は1、☓は0）＝0
    問題 8 （◯は1、☓は0）＝1
    回答＝10101001
    ```

<details><summary>▶プログラム例</summary>
    
```python
    x = 0
    for i in range(1, 9):
        ans = int(input(f"問題 {i} （◯は1、☓は0）＝"))
        x |= ans
        x <<= 1
    x >>= 1
    print(f"回答＝{x:08b}")
```

- 課題30

    正解の値を前もってセットしておく　`seikai = 0b11001110`
    
    課題29のように答えを変数`kotae`にセットし、回答入力後に採点して点数を表示する
    
    `saiten = seikai ^ kotae`のように排他的論理和(xor)を使用すると、正解と一致するビットが0、一致しないビットが1になる
    
    この値を0b11111111と排他的論理和(xor)をとると、正解のビットが1になるので、これを採点とする
    
    1の数を求めれば、点数になる
    
    （実行画面）
    
    ```
    問題 1 （◯は1、☓は0）＝1
    問題 2 （◯は1、☓は0）＝0
    問題 3 （◯は1、☓は0）＝1
    問題 4 （◯は1、☓は0）＝0
    問題 5 （◯は1、☓は0）＝1
    問題 6 （◯は1、☓は0）＝0
    問題 7 （◯は1、☓は0）＝0
    問題 8 （◯は1、☓は0）＝1
    回答＝10101001
    正解＝11001110
    採点＝10011000
    3点
    ```

<details><summary>▶プログラム例</summary>
    
```python
    seikai = 0b11001110
    kotae = 0
    for i in range(1, 9):
        ans = int(input(f"問題 {i} （◯は1、☓は0）＝"))
        kotae |= ans
        kotae <<= 1
    kotae >>= 1
    print(f"回答＝{kotae:08b}")
    print(f"正解＝{seikai:08b}")
    saiten = (seikai ^ kotae) ^ 0b11111111
    print(f"採点＝{saiten:08b}")
    cnt = 0
    for _ in range(8):
        if (saiten & 0b00000001) == 1:
            cnt += 1
        saiten >>= 1
    print(f"{cnt}点")
```
    
</details>
