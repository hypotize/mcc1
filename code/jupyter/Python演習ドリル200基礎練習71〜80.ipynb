- 課題72

    キーボードから学生の点数を入力する。
    
    文字Qが入力されたら、入力の終了とする。
    
    終了後に合計点、最高点、最低点を表示せよ。
    
    sum関数、max関数、min関数を使用すること。
    
    （実行画面）
    
    ```
    合計点と最高点、最低点を求めます。
    注："Q"で入力終了。
    1番目の点数：100
    2番目の点数：80
    3番目の点数：70
    4番目の点数：60
    5番目の点数：90
    6番目の点数：Q
    合計は400点です。
    最高は100点です。
    最低は60点です。
    ```

<details><summary>▶プログラム例</summary>

```python
    print("合計点と最高点、最低点を求めます。")
    print('注："Q"で入力終了。')
    score = []
    i = 1
    while True:
        s = input(f"{i}番の点数：")
        if s == "Q":
            break
        score.append(int(s))
        i += 1
    print(f"合計は{sum(score)}点です。")
    print(f"最高は{max(score)}点です。")
    print(f"最低は{min(score)}点です。")
        
```
    
</details>

- 課題74

    6人の2科目（国語、数学）の点数を乱数で読み込んで、
    
    科目ごとの平均値と学生ごと平均点を求める。
    
     （実行画面）
    
    ```
     生徒  国語 数学  平均
    ------------------------
    生徒1     0    99  49.5
    生徒2    69    47  58.0
    生徒3    82    75  78.5
    生徒4    54    63  58.5
    生徒5    69    71  70.0
    生徒6    99    58  78.5
    ------------------------
     平均  62.2  68.8
    ```   

<details><summary>▶プログラム例</summary>
    
```python
    import random

    students = []
    for _ in range(6):
        japanese = random.randint(0, 100)
        math = random.randint(0, 100)
        students.append((japanese, math))

    print(" 生徒  国語 数学  平均")
    print("------------------------")
    for i, score in enumerate(students):
        print(f"生徒{i+1} {score[0]:>5d} {score[1]:>5d} {sum(score)/len(score):>5.1f}")
    print("------------------------")
    print(f" 平均 {sum([score[0] for score in students])/len(students):>5.1f} {sum([score[1] for score in students])/len(students):>5.1f}")
```

- 課題75

    生徒の科目の点数を乱数で読み込んで、
    
    科目ごとの平均値と学生ごとの平均点を求める。
    
    生徒数、科目数が増えても、プログラム修正箇所が最低限に
    
    収まるようにせよ。
    
    （実行画面）
    
    ```
     生徒  国語  数学  英語  平均
    -------------------------------
    生徒1    14     1    22   12.3
    生徒2    64    57    69   63.3
    生徒3    81    73    17   57.0
    生徒4    34    43    79   52.0
    生徒5    44    72    11   42.3
    生徒6    30    27    18   25.0
    生徒7    64     1    82   49.0
    -------------------------------
     平均  47.3  39.1  42.6
    ```       

<details><summary>▶プログラム例</summary>
    
```python
    import random

    N = 7
    subjects = ["国語", "数学", "英語"]
    students = []
    for _ in range(N):
        score = []
        for _ in range(len(subjects)):
            score.append(random.randint(0, 100))
        students.append(score)

    print(" 生徒", end=" ")
    for subject in subjects:
        print(f" {subject}", end=" ")
    print(" 平均")
    print("--------------" + "-"*6*len(subjects))
    for i, score in enumerate(students):
        print(f"生徒{i+1}", end=" ")
        for j in range(len(subjects)):
            print(f"{score[j]:>5d}", end=" ")
        print(f" {sum(score)/len(score):>5.1f}")
    print("--------------" + "-"*6*len(subjects))
    print(" 平均", end=" ")
    for i in range(len(subjects)):
        print(f"{sum([score[i] for score in students])/len(students):>5.1f}", end=" ")
    print()
```
    
</details>
