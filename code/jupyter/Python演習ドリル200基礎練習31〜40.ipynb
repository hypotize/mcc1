- 課題39

    入力した文字列がすべて英字であれば、大文字に変換せよ
    
    すべて英字でなければ、「英字以外が含まれています。」と表示せよ
    
    （実行画面）
    
    ```
    文字列txt：Abc001
    英字以外が含まれています。
    ```
    ```
    文字列txt：Abc
    ABC
    ```

<details><summary>▶プログラム例</summary>

```python
    s = input("文字列txt：")
    if s.isalpha():
        print(s.upper())
    else:
        print("英字以外が含まれています。")
``` 
    
</details>
