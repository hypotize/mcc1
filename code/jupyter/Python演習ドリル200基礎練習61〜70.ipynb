    課題61
    
        10個のNoneが入ったリストを用意し、乱数で発生した0から9の数字を10個それぞれリストに格納して表示せよ
        
        そのリストに、さらに乱数を10個発生させて追加し、表示せよ

        （実行画面）
        
    ```
    [0, 4, 0, 9, 0, 8, 3, 0, 6, 8]
    [0, 4, 0, 9, 0, 8, 3, 0, 6, 8, 4, 3, 6, 3, 1, 6, 9, 5, 5, 7]
    ```

<details><summary>▶プログラム例</summary>

```python
    import random
    
    l = [None]*10
    for i in range(10):
        l[i] = random.randint(0, 9)
    print(l)
    for _ in range(10):
        l.append(random.randint(0, 9))
    print(l)
```

- 課題65

    次のリストを生成せよ
    
    ```
    生成方法：リスト表記演算子[]
    リスト名：month
    要素：'Jan', 'Feb', 'Mar', 'Apr'
    ```
    ```
    生成方法：リスト表記演算子[]
    リスト名：month2
    要素：'May', 'Jun', 'Jly', 'Aug'
    ```
    
    リストmonthの後ろにリストmonth2を累算代入で加えよ
    
    さらに、monthの後ろに、次の4つの要素をappendメソッドで加えよ
    
    `'Sep', 'Oct', 'Nov', 'Dec'`
    
    （実行画面）
    
    ```
    ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jly', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
    ```

<details><summary>▶プログラム例</summary>
    
```python
    month = ['Jan', 'Feb', 'Mar', 'Apr']
    month2 = ['May', 'Jun', 'Jly', 'Aug']
    month += month2
    month.append('Sep')
    month.append('Oct')
    month.append('Nov')
    month.append('Dec')
    print(month)
```

- 課題66

    次のプログラムのコメント行の下にコードを書け
    
    ```python
    color = ['blue', 'red', 'green', 'pink']
    print(color)
    # insertメソッドでインデックス1に'yellow'を追加
    # ここにコードを記入
    print(color)
    # removeメソッドで'red'を削除
    # ここにコードを記入
    print(color)
    # popメソッドで末尾の要素を削除
    # ここにコードを記入
    print(color)
    # del文で'yellow'を削除
    # ここにコードを記入
    print(color)
    ```
    
    （実行画面）
    
    ```
    ['blue', 'red', 'green', 'pink']
    ['blue', 'yellow', 'red', 'green', 'pink']
    ['blue', 'yellow', 'green', 'pink']
    ['blue', 'yellow', 'green']
    ['blue', 'green']
    ```

<details><summary>▶プログラム例</summary>
    
```python
    color = ['blue', 'red', 'green', 'pink']
    print(color)
    # insertメソッドでインデックス1に'yellow'を追加
    color.insert(1, 'yellow')
    print(color)
    # removeメソッドで'red'を削除
    color.remove('red')
    print(color)
    # popメソッドで末尾の要素を削除
    color.pop()
    print(color)
    # del文で'yellow'を削除
    del color[1]
    print(color)
```
