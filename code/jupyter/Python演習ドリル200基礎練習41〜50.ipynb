- 課題44

    文字列操作　しりとりゲーム
    
    実行例のように、
    
    ひらがなでしりとりができるプログラムを作成せよ
    
    入力した言葉がしりとりになっていないときは、入れ直し
    
    言葉の最後が'ん'のときは、終了
    
    （実行画面）
    
    ```
    最初の言葉を入れて＝りんご
    次の言葉をいれて＝ごりら
    次の言葉をいれて＝らっぱ
    次の言葉をいれて＝ぱり
    次の言葉をいれて＝さる
    入れ直してください。
    次の言葉をいれて＝かん
    最後が'ん'です
    ```

<details><summary>▶プログラム例</summary>
    
```python
    word = input("最初の言葉を入れて＝")
    while True:
        s = input("次の言葉を入れて＝")
        if s[-1] == 'ん':
            print("最後が'ん'です。")
            break
        elif s[0] == word[-1]:
            word = s
        else:
            print("入れ直してください。")
```
    
</details>

- 課題45

    単語の前に複数の空白がある文字列をセット　`s = 'This    is      a    pen   .'`
    
    単語の間の空白は1個にして表示せよ
    
    ただし、ピリオド（.)の前の空白は0個（なし）にせよ
    
    ヒント：　
    
    `split()`を使って余計な空白を除いて、単語とピリオドだけの配列にし、
    
    `join()`で改めて空白1個区切りの文字列にして、
    
    `replace()`で空白付きのピリオド（.）を空白なしのピリオド（.）に置き換える
    
    （実行画面）
    
    ```
    This    is      a    pen   .
    文字数＝26
    This is a pen.
    文字数＝14
    ```

<details><summary>▶プログラム例</summary>
    
```python
    s = 'This    is      a    pen   .'
    print(s)
    print(f"文字数＝{len(s)}")
    s = " ".join(s.split()).replace(' .', '.')
    print(s)
    print(f"文字数＝{len(s)}")    
```
    
</details>
