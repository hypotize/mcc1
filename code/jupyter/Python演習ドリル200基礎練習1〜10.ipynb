- 課題7

    商品の単価と数量を入力して、金額と消費税(10%)込みの金額を表示せよ
    
    print関数は1回だけ、f"文字列"を使用すること（文字列の中で{変数}が使えます）
    
    消費税(10%=0.1)はTAXという定数変数に設定しておくこと
    
    （実行画面）
    
```
    単価を入れてください=100
    数量を入れてください=10
    金額：1000
    消費税：100.0
    税込み金額：1100.0
```

<details><summary>▶プログラム例</summary>
    
```python
    TAX = 0.1
    price = int(input("単価を入力してください="))
    num = int(input("数量を入力してください"))
    amount = price * num
    print(f"金額：{amount}")
    tax = amount * TAX
    print(f"消費税：{tax}")
    total = tax + amount
    print(f"税込み金額：{total}")
```
    
</details>

3. 条件分岐

    - 課題8
    
        テストの点数を入力し、合格点（80点）以上であれば「合格」、
        
        そうでなければ「不合格」と表示せよ
        
        合格点は定数変数に設定しておくこと
        
        （実行結果）
        
```
    点数を入れてください=70
    不合格
```
```
    点数を入れてください=80
    合格
```

<details><summary>▶プログラム例</summary>
    
```python
    PASS_SCORE = 80
    score = int(input("点数を入れてください="))
    if score >= PASS_SCORE:
        print("合格")
    else:
        print("不合格")
```
    
</details>

- 課題9

    正の整数を入力し、偶数であれば「偶数です」、
    
    奇数であれば「奇数です」と表示せよ。
    
    マイナスの値が入力されたら、「正の整数ではありません」と表示せよ。
    
    （実行画面）
    
```
    正の整数を入れてください=-10
    正の整数ではありません
```
```
    正の整数を入れてください=5
    奇数です
```
```
    正の整数を入れてください=8
    偶数です
```

<details><summary>▶プログラム例</summary>
    
```python
    n = int(input("正の整数を入れてください="))
    if n <= 0:
        print("正の整数ではありません")
    elif n % 2 == 0:
        print("偶数です")
    else:
        print("奇数です")
```
    
</details>

- 課題10

    テストの点数を入力し、80点以上なら「A」、79〜60点なら「B」、
    
    59〜50点なら「C」、49点以下なら「D」と表示せよ。
    
    （実行画面）
    
```
    点数を入れてください=90
    A
```
```
    点数を入れてください=79
    B
```
```
    点数を入れてください=50
    C
```
```
    点数を入れてください=49
    D
```

<details><summary>▶プログラム例</summary>
    
```python
    score = int(input("点数をいれてください="))
    if score >= 80:
        print("A")
    elif score >= 60:
        print("B")
    elif score >= 50:
        print("C")
    else:
        print("D")
```
    
</details>
