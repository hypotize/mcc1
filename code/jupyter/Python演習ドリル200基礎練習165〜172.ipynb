        （実行結果）

<details><summary>▶プログラム例</summary>
    
```python
    import tkinter as tk
    
    def keisan():
        try:
            a = int(txtbox1.get())
            b = int(txtbox2.get())
        except Exception:
            lbl3.configure(text="数字を入力してください")
        else:
            lbl3.configure(text=f"かけ算の答えは {a*b} です")

    root = tk.Tk()

    root.geometry("500x450")  # ウィドウのサイズ　横x縦
    root.title("電卓アプリ")  # ウィンドウ名を設定
    
    lbl1 = tk.Label(text="数字を入力してください")
    lbl1.place(x=150, y=50)
    
    txtbox1 = tk.Entry(width=10)     # テクストボックスを作成（幅10）
    txtbox1.place(x=100, y=100)
    
    lbl2 = tk.Label(text="x")
    lbl2.place(x=210, y=100)
    
    txtbox2 = tk.Entry(width=10)     # テクストボックスを作成（幅10）
    txtbox2.place(x=250, y=100)
    
    lbl3 = tk.Label(text="かけ算の答えは")
    lbl3.place(x=180, y=250)    
    
    btn = tk.Button(text="計算", command=keisan)
    btn.place(x=200, y=300)
    
    tk.mainloop()
```
